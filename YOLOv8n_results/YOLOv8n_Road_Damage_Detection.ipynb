```python
from pathlib import Path
from collections import defaultdict
import hashlib

DATA_DIR = Path("/content/drive/MyDrive/road_data_shared")

hash_to_files = defaultdict(list)

for split in ["train", "val", "test"]:
    image_dir = DATA_DIR / "images" / split

    for image_path in image_dir.iterdir():
        if image_path.suffix.lower() not in {
            ".jpg", ".jpeg", ".png", ".bmp", ".webp"
        }:
            continue

        image_hash = hashlib.sha256(
            image_path.read_bytes()
        ).hexdigest()

        hash_to_files[image_hash].append(
            (split, image_path.name)
        )

duplicates = {
    image_hash: files
    for image_hash, files in hash_to_files.items()
    if len(files) > 1
}

cross_split_duplicates = {
    image_hash: files
    for image_hash, files in duplicates.items()
    if len({split for split, _ in files}) > 1
}

print("Total unique image hashes:", len(hash_to_files))
print("Exact duplicate groups:", len(duplicates))
print("Duplicates across different splits:", len(cross_split_duplicates))

for files in cross_split_duplicates.values():
    print(files)
```

```python
from pathlib import Path
from ultralytics import YOLO

# Step 1: Define the dataset configuration
DATA_YAML = Path(
    "/content/drive/MyDrive/road_data_shared/yolo_dataset/data.yaml"
)

# Step 2: Verify the configuration
assert DATA_YAML.exists(), f"Dataset configuration not found: {DATA_YAML}"

print("Dataset configuration verified!")

# Step 3: Load pretrained YOLOv8n
model = YOLO("yolov8n.pt")

# Step 4: Train the baseline model
results = model.train(
    data=str(DATA_YAML),
    epochs=50,
    imgsz=640,
    batch=16,
    device=0,
    seed=42,
    project="/content/runs",
    name="yolov8n_baseline",
    patience=10,
    workers=2,
    plots=True,
    save=True
)
```

```python
from ultralytics import YOLO
from pathlib import Path

DATA_YAML = "/content/drive/MyDrive/road_data_shared/yolo_dataset/data.yaml"

BEST_MODEL = "/content/runs/yolov8n_baseline/weights/best.pt"

assert Path(DATA_YAML).exists(), "data.yaml not found"
assert Path(BEST_MODEL).exists(), "best.pt not found"

# Load your trained YOLOv8n model
model = YOLO(BEST_MODEL)

# Evaluate on the test split
test_results = model.val(
    data=DATA_YAML,
    split="test",
    imgsz=640,
    batch=16,
    device=0,
    plots=True,
    project="/content/runs",
    name="yolov8n_test"
)

# Display final test metrics
print("\n===== YOLOv8n TEST RESULTS =====")
print("Precision:", round(test_results.box.mp, 4))
print("Recall:", round(test_results.box.mr, 4))
print("mAP@50:", round(test_results.box.map50, 4))
print("mAP@50-95:", round(test_results.box.map, 4))
```
