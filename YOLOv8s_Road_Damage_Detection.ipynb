```python
import os
from pathlib import Path

dataset_path = "/content/drive/MyDrive/Colab Notebooks/road_data_shared"

print("Folder exists:", os.path.exists(dataset_path))
print("Contents:", os.listdir(dataset_path))
```

```python
DATA_YAML = Path(dataset_path) / "yolo_dataset" / "data.yaml"

print("Config exists:", DATA_YAML.exists())
print(DATA_YAML.read_text())
```

```python

import yaml

config = yaml.safe_load(DATA_YAML.read_text())
config["path"] = dataset_path

MY_DATA_YAML = Path("/content/road_damage_yolov8s.yaml")
MY_DATA_YAML.write_text(
    yaml.safe_dump(config, sort_keys=False)
)

print(MY_DATA_YAML.read_text())
```

```python

from pathlib import Path

for split in ["train", "val", "test"]:
    image_dir = Path(dataset_path) / "images" / split
    label_dir = Path(dataset_path) / "labels" / split

    print(f"\n{split.upper()}")
    print("Images folder exists:", image_dir.exists())
    print("Labels folder exists:", label_dir.exists())

    if image_dir.exists():
        print("Number of images:", len(list(image_dir.glob("*"))))
    if label_dir.exists():
        print("Number of label files:", len(list(label_dir.glob("*.txt"))))
```

```python
from ultralytics import YOLO

model = YOLO("yolov8s.pt")

print("YOLOv8s pretrained model loaded successfully!")
```

```python
print("Dataset YAML:", MY_DATA_YAML)
print(MY_DATA_YAML.read_text())
```

```python
results = model.train(
    data=str(MY_DATA_YAML),
    epochs=50,
    imgsz=640,
    batch=16,
    device=0,
    seed=42,
    project="/content/runs",
    name="yolov8s_baseline",
    patience=10,
    workers=2,
    plots=True,
    save=True
)
```

```python

from ultralytics import YOLO
from pathlib import Path

BEST_MODEL = Path("/content/runs/yolov8s_baseline/weights/best.pt")

assert BEST_MODEL.exists(), "best.pt was not found!"

best_model = YOLO(str(BEST_MODEL))

test_results = best_model.val(
    data=str(MY_DATA_YAML),
    split="test",
    imgsz=640,
    batch=16,
    device=0,
    plots=True,
    project="/content/runs",
    name="yolov8s_test"
)

print("Test evaluation completed!")
```

```python

precision = test_results.box.mp
recall = test_results.box.mr
map50 = test_results.box.map50
map5095 = test_results.box.map

inference_ms = test_results.speed["inference"]
model_size_mb = BEST_MODEL.stat().st_size / (1024 * 1024)

print("===== YOLOv8s TEST RESULTS =====")
print(f"Precision:       {precision:.4f}")
print(f"Recall:          {recall:.4f}")
print(f"mAP@50:          {map50:.4f}")
print(f"mAP@50-95:       {map5095:.4f}")
print(f"Inference speed: {inference_ms:.2f} ms/image")
print(f"Model size:      {model_size_mb:.2f} MB")
```

```python

import pandas as pd
import shutil
from pathlib import Path

OUTPUT_DIR = Path(
    "/content/drive/MyDrive/Colab Notebooks/YOLOv8s_Results"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

metrics = pd.DataFrame([{
    "Model": "YOLOv8s",
    "Precision": precision,
    "Recall": recall,
    "mAP50": map50,
    "mAP50_95": map5095,
    "Inference_ms_per_image": inference_ms,
    "Model_size_MB": model_size_mb
}])

metrics.to_csv(OUTPUT_DIR / "test_metrics.csv", index=False)

shutil.copy2(BEST_MODEL, OUTPUT_DIR / "best.pt")

shutil.copy2(
    "/content/runs/yolov8s_baseline/results.csv",
    OUTPUT_DIR / "results.csv"
)

print("Saved files:")
for file in OUTPUT_DIR.iterdir():
    print(file.name)
```
